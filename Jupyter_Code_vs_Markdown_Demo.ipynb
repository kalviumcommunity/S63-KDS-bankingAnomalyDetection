# Code vs Markdown Cells Demo

Purpose: Show correct usage of Markdown cells for explanation and Code cells for execution.

## Explanation Before Code Cell 1

This code cell prints a message to verify Python execution in Jupyter.

```python
print("Hello from a Code cell")
```

## Output Explanation After Code Cell 1

The output should display:

`Hello from a Code cell`

This confirms the Code cell executed successfully.

## Explanation Before Code Cell 2

This code cell performs basic arithmetic and prints the results.

```python
a = 8
b = 4
print("Sum:", a + b)
print("Product:", a * b)
```

## Output Explanation After Code Cell 2

Expected output:

- `Sum: 12`
- `Product: 32`

### Code vs Markdown (Practical Use)

- Use **Code** cells to run Python and generate outputs.
- Use **Markdown** cells to add headings, explanations, and result interpretation.
- Keep notebooks structured as: explanation -> code -> explanation.
